```python
# Experiment: Read an image and convert it to grayscale

# Import required libraries
import cv2
import matplotlib.pyplot as plt

# Step 1: Specify the correct image path
# Replace 'your_image.jpg' with your actual image file name
image_path = '/content/Screenshot 2026-02-22 104814.png'

# Step 2: Read the image
image = cv2.imread(image_path)

# Step 3: Check if image is loaded properly
if image is None:
    print("Error: Image file not found.")
    print("Make sure the image is in the same folder as the program.")
    print("Or provide the full file path correctly.")
else:

    # Step 4: Convert original image from BGR to RGB
    original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Step 5: Convert image to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Step 6: Display images
    plt.figure(figsize=(10,5))

    # Original Image
    plt.subplot(1,2,1)
    plt.imshow(original_image)
    plt.title("Original Image")
    plt.axis("off")

    # Grayscale Image
    plt.subplot(1,2,2)
    plt.imshow(gray_image, cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    # Show output
    plt.show()

    # Step 7: Save grayscale image
    cv2.imwrite("grayscale_output.jpg", gray_image)

    print("Image successfully converted to grayscale.")
```

```python
# Experiment: Read an image and apply Gaussian Blur

# Import required libraries
import cv2
import matplotlib.pyplot as plt

# Step 1: Specify image path
image_path = '/content/W.jpg'

# Step 2: Read the image
image = cv2.imread(image_path)

# Step 3: Check if image is loaded successfully
if image is None:
    print("Error: Image file not found.")
else:

    # Step 4: Convert image from BGR to RGB for proper display
    original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Step 5: Apply Gaussian Blur
    # (5,5) is kernel size, 0 means automatic sigma calculation
    blurred_image = cv2.GaussianBlur(original_image, (5,5), 0)

    # Step 6: Display original and blurred images
    plt.figure(figsize=(10,5))

    # Original Image
    plt.subplot(1,2,1)
    plt.imshow(original_image)
    plt.title("Original Image")
    plt.axis("off")

    # Blurred Image
    plt.subplot(1,2,2)
    plt.imshow(blurred_image)
    plt.title("Gaussian Blurred Image")
    plt.axis("off")

    # Show images
    plt.show()

    # Step 7: Save blurred image
    blurred_bgr = cv2.cvtColor(blurred_image, cv2.COLOR_RGB2BGR)
    cv2.imwrite("gaussian_blur_output.jpg", blurred_bgr)

    print("Gaussian blurred image saved successfully.")
```

```python
# Experiment: Read an image and detect outlines using Canny Edge Detection

# Import required libraries
import cv2
import matplotlib.pyplot as plt

# Step 1: Specify image path
image_path = '/content/W.jpg'

# Step 2: Read the image
image = cv2.imread(image_path)

# Step 3: Check if image is loaded properly
if image is None:
    print("Error: Image file not found.")
else:

    # Step 4: Convert image to RGB for display
    original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Step 5: Convert image to grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Step 6: Apply Gaussian Blur to reduce noise
    blurred_image = cv2.GaussianBlur(gray_image, (5,5), 0)

    # Step 7: Apply Canny Edge Detection
    edges = cv2.Canny(blurred_image, 100, 200)

    # Step 8: Display original and outlined images
    plt.figure(figsize=(10,5))

    # Original Image
    plt.subplot(1,2,1)
    plt.imshow(original_image)
    plt.title("Original Image")
    plt.axis("off")

    # Edge Detected Image
    plt.subplot(1,2,2)
    plt.imshow(edges, cmap='gray')
    plt.title("Canny Outline Image")
    plt.axis("off")

    # Show output
    plt.show()

    # Step 9: Save edge-detected image
    cv2.imwrite("canny_outline_output.jpg", edges)

    print("Outline image saved successfully.")
```

```python
# Experiment: Histogram Equalization using OpenCV

# Import required libraries
import cv2
import matplotlib.pyplot as plt

# Step 1: Specify image path
image_path = '/content/JAS.jpg'

# Step 2: Read image in grayscale mode
image = cv2.imread(image_path, 0)

# Step 3: Check if image is loaded properly
if image is None:
    print("Error: Image file not found.")
else:

    # Step 4: Apply Histogram Equalization
    equalized_image = cv2.equalizeHist(image)

    # Step 5: Display original and equalized images
    plt.figure(figsize=(12,6))

    # Original Image
    plt.subplot(2,2,1)
    plt.imshow(image, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")

    # Equalized Image
    plt.subplot(2,2,2)
    plt.imshow(equalized_image, cmap='gray')
    plt.title("Histogram Equalized Image")
    plt.axis("off")

    # Original Histogram
    plt.subplot(2,2,3)
    plt.hist(image.ravel(), 256, [0,256])
    plt.title("Original Histogram")

    # Equalized Histogram
    plt.subplot(2,2,4)
    plt.hist(equalized_image.ravel(), 256, [0,256])
    plt.title("Equalized Histogram")

    # Show results
    plt.tight_layout()
    plt.show()

    # Step 6: Save equalized image
    cv2.imwrite("equalized_output.jpg", equalized_image)

    print("Histogram equalized image saved successfully.")
```

```python
# Experiment: Analyze histogram of an image based on color levels

# Import required libraries
import cv2
import matplotlib.pyplot as plt

# Step 1: Define histogram analysis function
def analyze_histogram(image_path):

    # Step 2: Read the image
    image = cv2.imread(image_path)

    # Step 3: Check if image is loaded properly
    if image is None:
        print("Error: Image file not found.")
        return

    # Step 4: Convert image from BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Step 5: Define colors for channels
    colors = ('b', 'g', 'r')

    # Step 6: Display original image
    plt.figure(figsize=(12,6))

    plt.subplot(1,2,1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Step 7: Plot histogram for each channel
    plt.subplot(1,2,2)

    for i, color in enumerate(colors):
        hist = cv2.calcHist([image], [i], None, [256], [0,256])
        plt.plot(hist, color=color)
        plt.xlim([0,256])

    plt.title("Color Histogram")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Number of Pixels")

    # Step 8: Show results
    plt.tight_layout()
    plt.show()

# Step 9: Call function
analyze_histogram('/content/L-IMAGE.jpg')
```
